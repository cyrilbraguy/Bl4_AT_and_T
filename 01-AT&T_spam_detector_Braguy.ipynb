# Load dataset 

```python
#load dataset file
import pandas as pd
max_cols = 5
column_names = [f"V{i+1}" for i in range(max_cols)]
dataset1 = pd.read_csv("./spam.csv",header=None, skiprows=1, names=column_names, encoding = "latin-1")
dataset1.head(21)
```

```python
dataset = pd.read_csv("./spam.csv",header=None, skiprows=1, names=column_names, encoding = "windows-1252")
dataset.head(21)
```

```python
dataset.info()
```

```python
dataset.rename(columns={"V1": "label", "V2": "message"}, inplace=True)
# sauve les index de ces lignes:
index_to_save = dataset[dataset["V3"].notna()].index    
dataset[dataset["V3"].notna()][["V3","V4","V5"]]
```

```python
# on ramene les morceaux de messages des colonnes V3, V4, V5 
# dans la cellule "message" de la meme ligne du dataset
dataset["message"] = dataset["message"].fillna("") + " " + dataset["V3"].fillna("") + " " + dataset["V4"].fillna("") + " " + dataset["V5"].fillna("")
dataset.loc[index_to_save, ["message","V3","V4","V5"]]
 
```

```python
# finaly clean columns V3, V4, V5 as they are now merged into "message"
dataset = dataset.drop(columns=["V3", "V4", "V5"])
dataset.head()
```

# Naive Bayes model to have a look 

```python
# Naive bayes model :
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report

X_train, X_test, y_train, y_test = train_test_split(
    dataset["message"].fillna("").astype(str),
    dataset["label"],
    test_size=0.2,
    random_state=42,
    stratify=dataset["label"]
)

model = make_pipeline(
    CountVectorizer(
        lowercase=True,
        token_pattern=r"(?u)\b\w+\b",
        ngram_range=(1, 2)
    ),
    MultinomialNB(alpha=1.0)
)

model.fit(X_train, y_train)
predictions = model.predict(X_test)

print(classification_report(y_test, predictions))
```

```python
from sklearn.metrics import f1_score
f1_score(y_test, predictions, average="weighted")
```

# model Deep learning simple : le LSTM Long Short Term Memory 

```python
import spacy

nlp = spacy.load("en_core_web_sm", disable=["parser", "ner"])


def clean_message(message: str) -> tuple[str, list[str], list[str]]:
    """Normalize an English SMS and return its text, tokens and lemmas.

    Punctuation is excluded from tokens and lemmas; numbers, URLs and stop words
    are retained because they can be useful for spam detection.
    """
    if not isinstance(message, str):
        raise TypeError("message must be a string")

    cleaned = " ".join(message.lower().split())
    doc = nlp(cleaned)
    words = [token for token in doc if not token.is_space and not token.is_punct]
    return cleaned, [token.text for token in words], [token.lemma_.lower() for token in words]


processed = dataset["message"].apply(clean_message)
dataset[["message_clean", "tokens", "lemmas"]] = pd.DataFrame(
    processed.tolist(), index=dataset.index
)
dataset[["label","message", "message_clean", "tokens", "lemmas"]].head()
```

## Vectorisation TF-IDF

On separe les SMS en jeux d'entrainement et de test avant d'apprendre le vocabulaire. Les matrices creuses obtenues peuvent ensuite servir a entrainer et evaluer un classificateur.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split

# TF-IDF attend des textes : on reunit les lemmes de chaque SMS.
texts = dataset["lemmas"].str.join(" ")
X_train_text, X_test_text, y_train, y_test = train_test_split(
    texts, dataset["label"], test_size=0.2, random_state=42,
    stratify=dataset["label"],
)

# Conserver les tokens spaCy, y compris les numeros et les mots d'une lettre.
vectorizer = TfidfVectorizer(tokenizer=str.split, token_pattern=None,
                             lowercase=False, ngram_range=(1, 2))
X_train = vectorizer.fit_transform(X_train_text)
X_test = vectorizer.transform(X_test_text)

print(f"Entrainement : {X_train.shape}, test : {X_test.shape}")
print(f"Nombre de termes : {len(vectorizer.get_feature_names_out())}")
```

## 1. Côté Technique : Les modèles d'IA pour détecter les spams SMS  
Si vous cherchez à entraîner ou intégrer un modèle de classification de texte (distinguer le "spam" du "ham" / message légitime), plusieurs approches se distinguent dans la recherche et le développement :  
• Naïve Bayes + BERT : L'association d'un classificateur traditionnel léger comme Naïve Bayes avec des plongements de mots (embeddings) issus de BERT offre un excellent compromis. Selon des publications de recherche récentes, cette combinaison atteint 97,31 % de précision avec un temps d'exécution ultra-rapide de 0,3 seconde.
• Spam Transformer Model : Un modèle modifié basé sur l'architecture des Transformers (publié dans IEEE Access) démontre une efficacité redoutable sur les jeux de données SMS de référence avec 98,92 % de précision.  
• Modèles classiques (Scikit-Learn) : Pour un projet Python plus léger et sans réseau de neurones lourd, le combo CountVectorizer / TF-IDF associé à une Régression Logistique, des Forêts Aléatoires (Random Forest) ou des machines à vecteurs de support (SVM) reste la solution standard de référence pour les développeurs.
