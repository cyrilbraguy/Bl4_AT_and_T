It is the world's largest telecommunications company by revenue and the third largest provider of mobile telephone services in the U.S. As of 2022, AT&T was ranked 13th on the Fortune 500 rankings of the largest United States corporations, with revenues of $168.8 billion! 😮

## Project 🚧

One of the main pain point that AT&T users are facing is constant exposure to SPAM messages.

AT&T has been able to manually flag spam messages for a time, but they are looking for an automated way of detecting spams to protect their users.

## Goals 🎯

Your goal is to build a spam detector, that can automatically flag spams as they come based solely on the sms' content.

## 1. Côté Technique : Les modèles d'IA pour détecter les spams SMS  
Si vous cherchez à entraîner ou intégrer un modèle de classification de texte (distinguer le "spam" du "ham" / message légitime), plusieurs approches se distinguent dans la recherche et le développement :  
• Naïve Bayes + BERT : L'association d'un classificateur traditionnel léger comme Naïve Bayes avec des plongements de mots (embeddings) issus de BERT offre un excellent compromis. Selon des publications de recherche récentes, cette combinaison atteint 97,31 % de précision avec un temps d'exécution ultra-rapide de 0,3 seconde.
• Spam Transformer Model : Un modèle modifié basé sur l'architecture des Transformers (publié dans IEEE Access) démontre une efficacité redoutable sur les jeux de données SMS de référence avec 98,92 % de précision.  
• Modèles classiques (Scikit-Learn) : Pour un projet Python plus léger et sans réseau de neurones lourd, le combo CountVectorizer / TF-IDF associé à une Régression Logistique, des Forêts Aléatoires (Random Forest) ou des machines à vecteurs de support (SVM) reste la solution standard de référence pour les développeurs.

## 2. Côté Utilisateur : Bloquer les spams sur votre téléphone
Si vous cherchez comment appliquer un modèle de protection efficace sur votre appareil mobile :
• Sur Android (Google Messages) : L'application intègre nativement un modèle de protection par IA. Allez dans Paramètres > ID de l'appelant et spam pour activer le filtrage automatique. Pour bloquer manuellement, faites un appui long sur la conversation pour la bloquer et la signaler comme spam.
• Sur iOS (iPhone) : Apple permet d'intégrer des extensions tierces via l'App Store, comme l'application française Anti SPAM SMS, pour filtrer automatiquement les messages promotionnels et le phishing avant qu'ils ne touchent votre boîte principale.  
• Le réflexe citoyen (France) : Quel que soit votre téléphone, vous pouvez transférer gratuitement le SMS suspect au 33700, la plateforme officielle de signalement qui transmet ces données aux opérateurs pour bloquer les réseaux de fraudeurs.  
Souhaitez-vous obtenir un exemple de code Python (Machine Learning) pour concevoir votre propre modèle de filtrage, ou cherchez-vous plutôt de l'aide pour configurer l'anti-spam sur un modèle de smartphone spécifique ?  

Alternatives si vous cherchez un modèle déjà prêt
• Modèles Open-Weight (Prêts à l'emploi) : Si vous n'avez pas encore votre propre fichier, vous pouvez récupérer des modèles d'architecture NLP (comme BERT ou RoBERTa) déjà fine-tunés sur des datasets de SMS spam directement sur la plateforme Hugging Face.
